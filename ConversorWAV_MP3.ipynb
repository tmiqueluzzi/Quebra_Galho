```python
import os

arquivos_wav = [
    f for f in os.listdir(".")
    if f.lower().endswith(".wav")
]

print(arquivos_wav)
```

```python
from pydub import AudioSegment
from pydub.utils import which

# Nessa lista cole o caminho do executável ffmpeg.exe
AudioSegment.converter = r"C:\Users\theom\Downloads\ffmpeg-2025-12-31-git-38e89fe502-full_build\ffmpeg-2025-12-31-git-38e89fe502-full_build\bin\ffmpeg.exe"

for musica in arquivos_wav[]:
    wav_path = musica
    mp3_path = musica[:-3] + 'mp3'

    audio = AudioSegment.from_wav(wav_path)
    audio.export(mp3_path, format="mp3", bitrate='320k')

print("Conversão concluída.")
```
